```python
def fahrenheit_to_celsius(fahrenheit_value):
    celsius_value=(fahrenheit_value-32)*(5/9)
    return celsius_value
```

```python
def celsius_to_kelvin(celsius_value):
    kelvin_value=celsius_value + 273.15
    return kelvin_value
def fahrenheit_to_kelvin(fahrenheit_value):
    celsius_value=fahrenheit_to_celsius(fahrenheit_value)
    kelvin_value=celsius_to_kelvin(celsius_value)
    return kelvin_value

print("freezing point of water in Kelvin:", fahrenheit_to_kelvin(32), "K")
```

```python
temp_K=fahrenheit_to_kelvin(32)
print(" the temperature in Kelvin is", temp_K)
```

```python
def reading_data(filename):
    """
    Purpose of the function: Read tabular data (.csv) from a file 
    Input: the path to the file to read (string)
    Output: The data in the file as a numpy array
    """
    data= np.loadtxt(filename, delimiter=",")

    return data
def analyze_data(data,axis_direction=0):

    maxval,minval,stdval,meanval=np.amax(data,axis=axis_direction),np.amin(data,axis=axis_direction),np.std(data,axis=axis_direction), np.mean(data, axis=axis_direction)

    return maxval, minval,stdval,meanval

def visualize_data(analyzed_data, axis_direction=0):

    fig=plt.figure(figsize=(15,3))

    axes1=fig.add_subplot(1,3,1)
    axes2=fig.add_subplot(1,3,2)
    axes3=fig.add_subplot(1,3,3)

    axes1.plot(analyzed_data[0])
    axes2.plot(analyzed_data[1])
    axes3.plot(analyzed_data[2])

    if axis_direction==0:
        for ax in [axes1,axes2,axes3]:
            ax.set_xlabel("Day")
    else:
        for ax in [axes1,axes2,axes3]:
            ax.set_xlabel("Patient")
    axes1.set_ylabel("mean")
    axes2.set_ylabel("max")
    axes3.set_ylabel("min")

    return fig
```

```python
import matplotlib.pylab as plt

file_pattern="data/inflammation-*.csv"
all_files=sorted(glob.glob(file_pattern))

for file in all_files:
    data=reading_data(file)
    maxval,minval,stdval,meanval=analyze_data(data)
    fig=visualize_data([meanval,maxval,minval])
    fig.suptitle(file,fontsize=16)

    
```
